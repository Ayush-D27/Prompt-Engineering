```python
from transformers import pipeline
import torch

device = 0 if torch.cuda.is_available() else -1

print("GPU Available:", torch.cuda.is_available())
```

```python
base_model = pipeline(
    "text-generation",
    model="distilgpt2",
    device=device
)

print("Base Model Loaded Successfully!")
```

```python
prompt = "Explain Artificial Intelligence in about 100 words."
```

```python
response = base_model(
    prompt,
    max_new_tokens=100,
    do_sample=True,
    temperature=0.7
)

print("="*60)
print("BASE MODEL OUTPUT")
print("="*60)

print(response[0]["generated_text"])
```
